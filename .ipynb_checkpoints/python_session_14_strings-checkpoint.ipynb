**MUTABLE-IMMUTABLE**
- Mutable means change
- Immutable means not able to change
- Example:'welcom'
- Outut : 'welcome'
- is this poossible using **index opreation**? no
- strings are **immutable**
- we can not change the letters using index

```python
str1='welcome'
str1[2]='L'
```

```python
string='hello how are you'
vowels='aeiou'
n=len(string)
count=0
for i in range(n):
    if string[i]=='aeiou':
        count+=1
print("the count of vowels is ",count)
```

**when to use range and when to use in**
- in operators acccess the elements directly
- range operators access the elements by using index
- for example :str='welcome'
- if you want to access each letters from str
- we can use in and range
- in:for i in str:i becomes w,e,l,c,o,m,e
- range:for in range(len(str)): i becomes 0 1 2 3 4 5
        -str[i]

**Use the range the operator when we you required a index use**

```python
str1='welcome'
for i in str1:
    print(i)
for i in range(len(str1)):
    print(str1[i])
```
